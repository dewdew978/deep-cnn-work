1. Load a Convolutional Neural Network

```python
from torchvision.models import resnet50, ResNet50_Weights
import torch.nn as nn
import torch

# New weights with accuracy 80.858%
model = resnet50()
model.fc = nn.Linear(2048, 1)
model.load_state_dict(torch.load('./age_model.pt'))
```

2. Test the network (One Image)

```python
import torchvision.transforms.v2 as transforms
import torch

transform = transforms.Compose([
    transforms.Resize((200, 200)),
    transforms.ToImage(),  # Convert to image format
    transforms.ConvertImageDtype(torch.float32),  # Convert image dtype
    transforms.Normalize(mean=(0.5, 0.5, 0.5), std=(0.5, 0.5, 0.5))])
```

```python
from PIL import Image
img_path = 'AgeDetection/30_1_2_20170116192249012.jpg'
image = Image.open(img_path).convert('RGB')
image = transform(image)

# Check if GPU (cuda) is available
device = torch.device('cuda:0' if torch.cuda.is_available() else 'cpu')

# Move our model to device (GPU or CPU)
model.to(device)
model.eval()

image = image.to(device)
image = image.unsqueeze(dim=0)

outputs = model(image)
predicted = outputs.squeeze(dim=0)
print(f'Predicted: {predicted.item():,.2f} years old', )
```
